```python
import requests
import pandas as pd

url = "https://jsonplaceholder.typicode.com/users"

users = requests.get(url).json()

data = []

for user in users:
    data.append({
        "id": user["id"],
        "name": user["name"],
        "email": user["email"],
        "city": user["address"]["city"]
    })

df = pd.DataFrame(data)

df.to_csv("users.csv", index=False)

print("Data saved successfully!")
```
